# $\Psi_1$

\begin{equation}
 \Psi_1(w;a)=\frac{1}{a}\frac{w}{w+1},
\quad\text{and}\quad \phi_1(w)=p-2- \frac{1}{\log(w+1)/a+1/(p-2)},
\end{equation}


```r
P = 3:15
A = (1:7)/2
```

```r
result.lower = matrix(0, length(A), length(P))
```

```r
for(i in 1:length(A)){
  for(k in 1:length(P)){
    denominator = function(x){
    dchisq(x,df= P[k]-2) /(1/(P[k]-2) + log(x+1)/A[i])^2
    }
    numerator = function(x){
      denominator(x) * x/(x+1)
    }
  In = integrate(numerator, 0, Inf)
  Id = integrate(denominator, 0, Inf)
  tmp = 4*(1/A[i])*(In$value-In$abs.error)/(Id$value+Id$abs.error)-1
  if(tmp<0){
    result.lower[i, k]=round(tmp,3)
  }else{
    result.lower[i, k]=round(tmp,3)
    break
    }
  }
}
```

```r
rownames(result.lower) = as.character(A)
colnames(result.lower) = as.character(P)
result.lower
```

# $\Psi_2$

\begin{equation}
 \Psi_2(w;a,b)=\frac{1}{a}\left(\frac{w}{p-2}\right)^b%
 \quad\text{for }a>0,\ b>0,
\quad \text{and}\quad\phi_2(w)=p-2- \frac{1}{ \{w/(p-2)\}^b/(ab)+1/(p-2)},
\end{equation}


```r
P = 3:300
A = (1:7)/2
B = (1:6)/2
```

```r
result.lower = array(0, dim=c(length(P),length(A), length(B)))
result.P = matrix(0, length(A),length(B))
```

```r
for(i in 1:length(A)){
  for(j in 1:length(B)){
    for(k in 1:length(P)){
      if(P[k]/2-1-2*B[j]>0){
        y = P[k]/2-1
        tmp = exp(lgamma(y-B[j])-lgamma(y-2*B[j])-B[j]*log(y)-log(A[i]))
#        tmp = gamma(y-B[j])/(gamma(y-2*B[j])*y^(B[j]))/A[i]
        result.lower[k,i,j]=round(4*tmp-1,2)
        if(4*tmp-1>0){
          if(result.P[i,j]==0){
            result.P[i,j]=P[k]
          }
        }
      }
    }
  }
}
```

```r
rownames(result.P) = as.character(A)
colnames(result.P) = as.character(B)
result.P
```

```r
result.PP.2 <- matrix(list(NULL), length(A), length(B))
result.PP.3 <- matrix(list(NULL), length(A), length(B))
```

```r
for(i in 1:length(A)){
  for(j in 1:length(B)){
    for(k in (result.P[i,j]):3){
      denominator = function(x){
        dgamma(x,shape=k/2-1, scale=1/(k/2-1))/(1/(k-2) + x^(B[j])/(A[i]*B[j]))^2
      }
      numerator = function(x){
         denominator(x) * x^(B[j])/A[i]
      }
    In = integrate(numerator, 0, Inf)
    Id = integrate(denominator, 0, Inf)

    tmp = 4*(In$value-In$abs.error)/(Id$value+Id$abs.error)-1
    result.PP.3[[i,j]] = c(result.PP.3[[i,j]], round(tmp,2))
    if(tmp>0){
    result.PP.2[[i,j]] = c(result.PP.2[[i,j]],k)
    }
    }
  }
}
```
